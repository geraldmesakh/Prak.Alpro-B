```python
def cek_prima(n, pembagi=2):
    if n < 2:
        return False

    if pembagi * pembagi > n:
        return True

    if n % pembagi == 0:
        return False

    return cek_prima(n, pembagi + 1)

bilangan = int(input("Masukkan bilangan: "))

if cek_prima(bilangan):
    print(bilangan, "adalah bilangan prima")
else:
    print(bilangan, "bukan bilangan prima")
```

```python
def palindrom(kalimat):
    kalimat = kalimat.lower().replace(" ", "")

    if len(kalimat) <= 1:
        return True

    if kalimat[0] != kalimat[-1]:
        return False

    return palindrome(kalimat[1:-1])

teks = input("Masukkan kalimat: ")

if palindrom(teks):
    print("Kalimat tersebut adalah palindrom")
else:
    print("Kalimat tersebut bukan palindrom")
```

```python
def jumlah_ganjil(n):
    if n == 1:
        return 1

    return n + jumlah_ganjil(n - 2)

n = int(input("Masukkan bilangan ganjil terakhir: "))

print("Jumlah deret ganjil =", jumlah_ganjil(n))
```

```python
def jumlah_digit(n):
    if n < 10:
        return n

    return (n % 10) + jumlah_digit(n // 10)

bilangan = int(input("Masukkan bilangan: "))

print("Jumlah digit =", jumlah_digit(bilangan))
```

```python
def faktorial(n):
    if n == 0 or n == 1:
        return 1
    return n * faktorial(n - 1)

def kombinasi(n, r):
    return faktorial(n) // (faktorial(r) * faktorial(n - r))

n = int(input("Masukkan nilai n: "))
r = int(input("Masukkan nilai r: "))

print("Nilai kombinasi =", kombinasi(n, r))
```
