```python
playstore = {
    "Game": {"Mobile Legends", "Free Fire", "PUBG"},
    "Sosial": {"WhatsApp", "Instagram", "Facebook"},
    "Produktivitas": {"Google Docs", "WhatsApp", "Zoom"},
    "Komunikasi": {"WhatsApp", "Telegram", "Zoom"}
}

jumlah_muncul = {}

for kategori in playstore:
    for aplikasi in playstore[kategori]:
        if aplikasi in jumlah_muncul:
            jumlah_muncul[aplikasi] += 1
        else:
            jumlah_muncul[aplikasi] = 1

print("Aplikasi yang muncul di satu kategori saja:")
for aplikasi in jumlah_muncul:
    if jumlah_muncul[aplikasi] == 1:
        print("-", aplikasi)

n = int(input("\nMasukkan nilai n: "))

if n > 2:
    print("\nAplikasi yang muncul tepat di dua kategori:")
    for aplikasi in jumlah_muncul:
        if jumlah_muncul[aplikasi] == 2:
            print("-", aplikasi)
```

```python
list_data = [1, 2, 3, 2, 4]
set_data = {5, 6, 7}
tuple_data = (8, 9, 10, 8)

print("List sebelum konversi :", list_data)
hasil_set = set(list_data)
print("Set sesudah konversi  :", hasil_set)

print()

print("Set sebelum konversi  :", set_data)
hasil_list = list(set_data)
print("List sesudah konversi :", hasil_list)

print()

print("Tuple sebelum konversi :", tuple_data)
hasil_set2 = set(tuple_data)
print("Set sesudah konversi   :", hasil_set2)

print()

print("Set sebelum konversi   :", set_data)
hasil_tuple = tuple(set_data)
print("Tuple sesudah konversi :", hasil_tuple)
```

```python
try:
    file1 = input("Masukkan nama file pertama : ")
    file2 = input("Masukkan nama file kedua   : ")

    f1 = open(file1, "r")
    isi1 = f1.read().lower()
    f1.close()

    f2 = open(file2, "r")
    isi2 = f2.read().lower()
    f2.close()

    kata1 = set(isi1.split())
    kata2 = set(isi2.split())

    sama = kata1.intersection(kata2)

    print("\nKata yang muncul pada kedua file:")
    
    if len(sama) > 0:
        for kata in sama:
            print("-", kata)
    else:
        print("Tidak ada kata yang sama")

except FileNotFoundError:
    print("Error: File tidak ditemukan atau tidak bisa dibaca")
```
